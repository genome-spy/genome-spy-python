# Connect chart interactions to Python

Keep `component.py` beside this notebook. Run all cells, wait for **Ready**, then drag across the chart and release. Python adds the selected region to a list and updates the lower track. Nothing is written to a file; rerunning the setup cell clears the list. Coordinates are fictional, not a real genome.

````python
import asyncio
import html

import ipywidgets as widgets
from IPython.display import display

from component import annotate, chart

# Rerunning starts a fresh session.
if old_task := globals().get("connection_task"):
    old_task.cancel()
if old_widget := globals().get("widget"):
    old_widget.close()

annotations = []
widget = chart.widget(inline=True, controls=False)
status = widgets.HTML("Connecting to Python…")
display(widget, status)


def show(message):
    status.value = html.escape(message)


async def connect():
    try:
        async with asyncio.timeout(30):
            api = await widget.get_embed_api()
        await annotate(api, annotations, show)
    except Exception as error:
        show(f"{type(error).__name__}: {error}")


# Keep listening for selections without leaving this cell running.
connection_task = asyncio.create_task(connect())
````

Run this cell after brushing to inspect the Python list. Rerun it to refresh its output. The status above updates automatically. Records stay in memory until you save them to a file.
