# Basic GenomeSpy Notebook Example

This notebook renders a simple GenomeSpy chart through the project's `anywidget` integration.

```python
import genome_spy as gs
```

```python
chart = (
    gs.Chart(
        data=[
            {"x": 1, "y": 2, "category": "A"},
            {"x": 2, "y": 3, "category": "B"},
            {"x": 3, "y": 1, "category": "C"},
        ]
    )
    .mark_point(size=90)
    .encode(
        x=gs.X("x:Q"),
        y=gs.Y("y:Q"),
        color=gs.Color("category:N"),
    )
    .properties(
        width=420,
        height=220,
        description="A minimal GenomeSpy point chart rendered in a notebook.",
    )
)
```

```python
chart
```
