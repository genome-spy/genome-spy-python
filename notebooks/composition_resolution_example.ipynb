# Composition Resolution Example

This notebook demonstrates a real concat composition using `left | right`.
GenomeSpy is strict about parent-level `scales`, `axes`, and `legends` on
concat roots when they would map to multiple visible resolutions, so this
example keeps the concat valid and shows the resulting side-by-side plot.


```python
import genome_spy as gs

left_data = [
    {"position": 1, "value": 2.0, "species": "setosa"},
    {"position": 2, "value": 3.5, "species": "versicolor"},
    {"position": 3, "value": 2.8, "species": "virginica"},
]

right_data = [
    {"position": 1, "score": 4.0, "species": "setosa"},
    {"position": 2, "score": 3.2, "species": "versicolor"},
    {"position": 3, "score": 4.4, "species": "virginica"},
]
```

```python
left = (
    gs.Chart(left_data)
    .mark_point(size=90)
    .encode(
        x=gs.X("position:Q"),
        y=gs.Y("value:Q").scale(reverse=False),
        color=gs.Color("species:N"),
    )
    .properties(width=220, height=240, title="Left panel")
)

right = (
    gs.Chart(right_data)
    .mark_point(size=90)
    .encode(
        x=gs.X("position:Q"),
        y=gs.Y("score:Q").scale(reverse=False),
        color=gs.Color("species:N"),
    )
    .properties(width=220, height=240, title="Right panel")
)

chart = (left | right).properties(title="Concat composition")

chart
```
