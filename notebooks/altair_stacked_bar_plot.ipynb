# GenomeSpy-Style Stacked Bar Chart

GenomeSpy does not use Vega-Lite's encoding-level aggregate shorthand for this pattern. Instead, the spec follows GenomeSpy's own stacked-bar example: aggregate the data first, stack the aggregated values, then encode the generated start/end fields.

```python
import genome_spy as alt
from altair.datasets import data

source = data.barley()

chart = (
    alt.Chart(source)
    .transform_aggregate(
        groupby=["variety", "site"],
        fields=["yield"],
        ops=["sum"],
        as_=["yieldSum"],
    )
    .transform_stack(
        field="yieldSum",
        groupby=["variety"],
        sort={"field": "site", "order": "ascending"},
        as_=["yieldStart", "yieldEnd"],
    )
    .mark_rect()
    .encode(
        x=alt.X("yieldStart:Q").title("Sum of yield"),
        x2="yieldEnd",
        y=alt.Y("variety:N").scale(padding=0.1, reverse=False),
        color="site:N",
    )
    .properties(width=520, height=340, title="Barley Yield by Variety and Site")
)

chart
```
