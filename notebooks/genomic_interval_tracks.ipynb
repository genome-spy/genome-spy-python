# Genomic Interval Tracks

This notebook is a small GenomeSpy-native genomics example built from synthetic data.
It composes an interval track over a variant-level signal track using locus encodings,
secondary `x2`, and vertical concatenation.

```python
import genome_spy as gs

intervals = [
    {"chrom": "chr1", "start": 100_000, "end": 180_000, "feature": "promoter"},
    {"chrom": "chr3", "start": 24_200_000, "end": 24_320_000, "feature": "exon"},
    {"chrom": "chr7", "start": 55_100_000, "end": 55_260_000, "feature": "enhancer"},
    {"chrom": "chr12", "start": 11_800_000, "end": 11_940_000, "feature": "promoter"},
    {"chrom": "chr17", "start": 43_050_000, "end": 43_210_000, "feature": "exon"},
    {"chrom": "chr22", "start": 36_400_000, "end": 36_520_000, "feature": "enhancer"},
]

variants = [
    {"chrom": "chr1", "pos": 120_000, "signal": 4.2, "impact": "moderate"},
    {"chrom": "chr2", "pos": 91_400_000, "signal": 7.1, "impact": "high"},
    {"chrom": "chr3", "pos": 24_260_000, "signal": 5.6, "impact": "low"},
    {"chrom": "chr5", "pos": 12_700_000, "signal": 6.4, "impact": "moderate"},
    {"chrom": "chr7", "pos": 55_180_000, "signal": 8.4, "impact": "high"},
    {"chrom": "chr9", "pos": 68_500_000, "signal": 3.9, "impact": "low"},
    {"chrom": "chr12", "pos": 11_880_000, "signal": 5.1, "impact": "moderate"},
    {"chrom": "chr17", "pos": 43_160_000, "signal": 7.8, "impact": "high"},
    {"chrom": "chr22", "pos": 36_470_000, "signal": 4.8, "impact": "low"},
]
```

```python
interval_track = (
    gs.Chart(intervals)
    .mark_rect(cornerRadius=3)
    .encode(
        x=gs.Locus("chrom", "start", scale={"assembly": "hg38"}),
        x2=gs.Locus("chrom", "end"),
        y=gs.Y("feature:N").scale(reverse=False, padding=0.25),
        color=gs.Color("feature:N").legend(title="Interval type"),
    )
    .properties(height=120)
)

signal_track = (
    gs.Chart(variants)
    .mark_point(filled=True, size=110)
    .encode(
        x=gs.Locus(
            "chrom",
            "pos",
            scale={"assembly": "hg38"},
            axis={"title": "Genomic position"},
        ),
        y=gs.Y("signal:Q").scale(reverse=False, zero=True),
        color=gs.Color("impact:N").legend(title="Impact"),
        size=gs.Size("signal:Q").scale(zero=False),
    )
    .properties(height=220)
)

chart = interval_track & signal_track
chart
```
