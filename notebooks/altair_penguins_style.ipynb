# Altair-Style Penguins Example

This notebook intentionally follows the shape of an Altair example. The dataset still comes from `altair.datasets`, while the chart object comes from `genome_spy`.

```python
import genome_spy as alt
from altair.datasets import data

source = data.penguins()
source.head()
```

```python
data.penguins()
```

```python
chart = (
    alt.Chart(source)
    .mark_circle()
    .encode(
        alt.X("Flipper Length (mm)").scale(zero=False),
        alt.Y("Body Mass (g)").scale(zero=False, padding=1),
        alt.Size("Beak Depth (mm)").scale(zero=False),
        color="Species",
    )
    .properties(width=520, height=340, title="Palmer Penguins", theme="vegalite")
)

chart
```
