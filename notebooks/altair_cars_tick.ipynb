# Altair-Style Cars Tick Chart

This notebook mirrors a small Altair example while importing `genome_spy` as `alt`.

```python
import genome_spy as alt
from altair.datasets import data

source = data.cars()

chart = (
    alt.Chart(source)
    .mark_tick()
    .encode(
        x="Horsepower:Q",
        y="Cylinders:O",
    )
)

chart
```
