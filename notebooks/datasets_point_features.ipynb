# Built-in Point Feature Dataset

This notebook uses the small `point_features` dataset bundled with `genome_spy`.

```python
import genome_spy as gs
from genome_spy.datasets import data

data.list_datasets()
```

```python
features = data.point_features()
features[:2]
```

```python
chart = (
    gs.Chart(data=features)
    .mark_point(size=120)
    .encode(
        x=gs.X("pos:Q", title="Position"),
        y=gs.Y("score:Q", title="Score"),
        color=gs.Color("category:N", title="Category"),
    )
    .properties(width=520, height=260, title="Toy Genomic Point Features")
)

chart
```
