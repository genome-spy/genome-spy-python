# Layered Resolution Helpers Example

This notebook demonstrates composition-level `resolve_scale(...)`,
`resolve_axis(...)`, and `resolve_legend(...)` on a schema-backed layered
chart. The example avoids channels where child encodings already carry
their own scale config, because GenomeSpy rejects mixing view-level and
encoding-level scale settings in the same resolution.


```python
import genome_spy as gs
from genome_spy.schema import GenomeAxis, Legend, Scale

source = [
    {"position": 1, "value": 2.0, "species": "setosa", "label": "S1"},
    {"position": 2, "value": 3.5, "species": "versicolor", "label": "V2"},
    {"position": 3, "value": 2.8, "species": "virginica", "label": "V3"},
]
```

```python
points = (
    gs.Chart(source)
    .mark_point(size=100)
    .encode(
        x=gs.X("position:Q"),
        y=gs.Y("value:Q"),
        color=gs.Color("species:N"),
    )
)

labels = (
    gs.Chart(source)
    .mark_text(dy=-12)
    .encode(
        x=gs.X("position:Q"),
        y=gs.Y("value:Q"),
        text=gs.Text("label:N"),
        color=gs.Color("species:N"),
    )
)

chart = (
    (points + labels)
    .resolve_scale(
        x=Scale(reverse=True),
        color=Scale(scheme="blues"),
    )
    .resolve_axis(x=GenomeAxis(title="Position"))
    .resolve_legend(color=Legend(title="Species"))
    .properties(
        width=420,
        height=260,
        title="Layered composition-level resolution helpers",
    )
)

chart
```
