# Built-in Copy Number Segment Dataset

This notebook uses the small `copy_number_segments` dataset bundled with `genome_spy`.

```python
import genome_spy as gs
from genome_spy.datasets import data

segments = data.copy_number_segments()
segments[:2]
```

```python
chart = (
    gs.Chart(data=segments)
    .mark_rect(opacity=0.9)
    .encode(
        x=gs.X("start:Q", title="Segment start"),
        x2=gs.X2("end:Q"),
        y=gs.Y("sample:N", title="Sample"),
        color=gs.Color("copy_number:Q", title="Copy number"),
    )
    .properties(width=620, height=220, title="Toy Copy Number Segments")
)

chart
```
