# Ranked genes

Plot gene rankings already stored by Scanpy. The packaged PBMC data contains rankings for the cell types used here; these examples do not rerun differential expression.

Unreleased preview for Scanpy 1.11.5. Install with `pip install -e '.[scanpy]'` from the project folder and use that environment as your Jupyter kernel.


```python
import scanpy as sc
import genome_spy.integrations.scanpy as gsc

pbmc = sc.datasets.pbmc68k_reduced()
ranked_groups = ["CD19+ B", "CD14+ Monocyte"]
sc.tl.dendrogram(pbmc, groupby="bulk_labels")
```

## Inspect ranking scores

Show the five highest-ranked genes for each selected cell type. Hover for the stored scores.


```python
scores = gsc.pl.rank_genes_groups(
    pbmc,
    groups=ranked_groups,
    n_genes=5,
    ncols=2,
    width=240,
    height=280,
    show=False,
)
scores
```

## Summarize ranked genes with dots

Use the top three genes from each selected ranking. Color shows mean expression; size shows the fraction of cells expressing each gene.


```python
ranked_dots = gsc.pl.rank_genes_groups_dotplot(
    pbmc,
    groups=ranked_groups,
    n_genes=3,
    zoom=True,
    highlight=True,
    width=240,
    height=320,
    show=False,
)
ranked_dots
```

## Compare ranked gene means

Use the same ranked genes in a matrixplot. Each gene’s group means are scaled from 0 to 1.


```python
ranked_means = gsc.pl.rank_genes_groups_matrixplot(
    pbmc,
    groups=ranked_groups,
    n_genes=3,
    standard_scale="var",
    zoom=True,
    width=240,
    height=320,
    show=False,
)
ranked_means
```

## Inspect ranked genes in individual cells

Show individual cells for the same ranked gene set. Scroll to zoom rows and hover for cell IDs and values.


```python
ranked_cells = gsc.pl.rank_genes_groups_heatmap(
    pbmc,
    groups=ranked_groups,
    n_genes=3,
    standard_scale="var",
    cmap="viridis",
    zoom=True,
    width=300,
    height=400,
    show=False,
)
ranked_cells
```
