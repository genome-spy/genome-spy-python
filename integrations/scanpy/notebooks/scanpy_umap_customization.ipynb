# Scanpy advanced plotting tutorial: UMAP customization

Adapt the supported UMAP sections of Scanpy’s [advanced plotting tutorial](https://scanpy.scverse.org/en/stable/tutorials/plotting/advanced.html) to GenomeSpy. Use the tutorial’s real PBMC dataset and stored embedding. Scanpy performs any explicit analysis; GenomeSpy performs rendering and interaction.

This notebook covers panel layout, subsetting, titles, category order, categorical palettes, and continuous color limits. Matplotlib styling chains, custom normalizers, and the tutorial’s other plot families are outside the current adapter support.

Colors use GenomeSpy schemes and opaque CSS colors. This example uses native `redblue` in place of Matplotlib’s `coolwarm`.

```python
import pandas as pd
import scanpy as sc

import genome_spy.scanpy as gsc

pbmc = sc.datasets.pbmc68k_reduced()
```

## Side-by-side panels

Replace the tutorial’s `sc.pl.umap` with `gsc.pl.umap` and set panel dimensions directly.

```python
chart = gsc.pl.umap(
    pbmc,
    color=["louvain", "phase"],
    wspace=1,
    width=280,
    height=280,
    show=False,
)
chart.display(inline=True, controls=False)
```

## A subset retains its supplied coordinates

Use the tutorial’s Louvain groups 0, 3, and 9, colored by S-phase score. This is an AnnData view of the existing embedding; no new UMAP is computed.

```python
subset = pbmc[pbmc.obs["louvain"].isin(["0", "3", "9"])]
chart = gsc.pl.umap(subset, color="S_score", width=280, height=280, show=False)
chart.display(inline=True, controls=False)
```

## Title and frame

```python
chart = gsc.pl.umap(
    pbmc,
    color="bulk_labels",
    title="Cell type",
    frameon=False,
    width=280,
    height=280,
    show=False,
)
chart.display(inline=True, controls=False)
```

## Categorical order

Create an explicitly ordered annotation in Python, as in the tutorial. Legend order follows the categorical levels.

```python
pbmc.obs["phase_ordered"] = pd.Categorical(
    pbmc.obs["phase"],
    categories=["G1", "S", "G2M"],
    ordered=True,
)
chart = gsc.pl.umap(
    pbmc,
    color=["phase", "phase_ordered"],
    wspace=0.5,
    width=280,
    height=280,
    show=False,
)
chart.display(inline=True, controls=False)
```

## An explicit categorical palette

```python
chart = gsc.pl.umap(
    pbmc,
    color="phase",
    s=20,
    palette={"S": "#17becf", "G1": "#bcbd22", "G2M": "#d62728"},
    width=280,
    height=280,
    show=False,
)
chart.display(inline=True, controls=False)
```

## A real gene score with a symmetric color range

The advanced tutorial scores the B-cell markers CD79A and MS4A1 with Scanpy. That explicit analysis adds `B_cell_score` to `obs`; the plotting call only reads it.

```python
sc.tl.score_genes(pbmc, ["CD79A", "MS4A1"], score_name="B_cell_score", random_state=0)
maxabs = abs(pbmc.obs["B_cell_score"]).max()
chart = gsc.pl.umap(
    pbmc,
    color="B_cell_score",
    cmap="redblue",
    s=20,
    vmin=-maxabs,
    vmax=maxabs,
    width=280,
    height=280,
    show=False,
)
chart.display(inline=True, controls=False)
```

Data: `scanpy.datasets.pbmc68k_reduced()` provides Scanpy’s packaged, processed subset of 10x Genomics PBMC68k. Category reordering and gene scoring above are explicit notebook operations; the original UMAP coordinates are retained throughout.
